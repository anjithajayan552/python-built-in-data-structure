PART A - WARM UP PROGRAMS

A1.  favorite fruits

```python
fruits = ["apple", "banana", "grapes", "mango", "orange"]

print("First:", fruits[0])
print("Last:", fruits[-1])
print("Count:", len(fruits))
```

A2. Mark Summary

```python
marks = [78, 92, 65, 88, 70]

print("Total:", sum(marks))
print("Average: {:.2f}".format(sum(marks) / len(marks)))
print("Highest:", max(marks))
print("Lowest:", min(marks))
```

A3. read number into a list

```python
numbers = []

for i in range(1, 6):
    num = int(input("Enter number {}: ".format(i)))
    numbers.append(num)

print(numbers)
```

A4. numbered list

```python
colours = ["red", "green", "blue", "yellow"]

for i, colour in enumerate(colours, start=1):
    print(f"{i}. {colour}")
```

A5. change a list step by step

```python
colours = ["red", "green", "blue", "yellow"]

for i, colour in enumerate(colours, start=1):
    print(f"{i}. {colour}")
```

A6. Sorted copies

```python
nums = [3, 8, 1, 9, 4]

print(sorted(nums))
print(sorted(nums, reverse=True))
print(nums)
```

A7. search a value

```python
nums = [15, 42, 8, 23, 42]

number = int(input("Enter a number: "))

if number in nums:
    print("Found at index", nums.index(number))
else:
    print("Not found")
```

A8 . first three and last three

```python
numbers = list(range(1, 11))

print(numbers[:3])
print(numbers[-3:])
```

A9. Count occurences

```python
numbers = [2, 5, 2, 7, 2, 9]

print(numbers.count(2))

count = 0

for num in numbers:
    if num == 2:
        count += 1

print(count)
```

A10. three ways to reverse

```python
numbers = [1, 2, 3, 4, 5]

# Method 1: Slicing
print(numbers[::-1])

# Method 2: reverse() method
nums2 = numbers.copy()
nums2.reverse()
print(nums2)

# Method 3: reversed() function
print(list(reversed(numbers)))
```
